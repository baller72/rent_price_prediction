```python
# Importation des librairies
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Importation du dataset
df = pd.read_csv("../data/processed_dataset.csv")
df.head()
```

# Créer *Confort_Score*, puis supprimer les 5 colonnes individuelles.

Le calcul de **Confort_Score** ne doit pas être réalisé directement sur des variables contenant des valeurs manquantes, car cela produirait un score partiellement biaisé. Les variables binaires doivent d’abord être encodées puis imputées avant la création de cette variable composite, afin de garantir une interprétation cohérente du niveau de confort du logement.

```python
# Créer une variable Confort_Score correspondant à la somme des 5 variables binaires
binary_cols = ["Salon", "SalleDeBainInterieure", "Parking", "Meuble", "Jardin"]

# Remplacement des textes avec valeurs
for col in binary_cols:
    df[col] = df[col].map({"Oui": 1, "Non": 0})

# Imputation avant score
for col in binary_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

df["Confort_Score"] = df[binary_cols].sum(axis=1)
#df['Confort_Score'] = df['Salon'] + df['SalleDeBainInterieure'] + df['Parking'] + df['Meuble'] + df['Jardin']
```

```python
# Supprimer ensuite les 5 colonnes individuelles
df = df.drop(columns=["Salon", "SalleDeBainInterieure", "Parking", "Meuble", "Jardin"], axis=1)
df.head()
```

# Création de *Chambres_par_Superficie* 

```python
# Création de la feature de densité (AVANT de supprimer la colonne Chambres)
df['Chambres_par_Superficie'] = df['Chambres'] / df['Superficie_m2']
```

```python
df.to_csv("../data/train_data.csv", index=False)
df
```
